# 3: Collections

## List
Mainly used when items are ordered and allows for duplicates


```kotlin
// Read-only & mutable list(with type declaration)
val readOnlyShapes = listOf("triangle", "square", "circle")
println(readOnlyShapes)

val shapes: MutableList<String> = mutableListOf("triangle", "square", "circle")
println(shapes)
```

```kotlin
// Casting(prevent unwanted modifications to list)
val shapes: MutableList<String> = mutableListOf("triangle", "square", "circle")
val shapesLocked: List<String> = shapes

shapes.add("hexagon") // Original still modifiable
shapesLocked.add("pentagon") // Prevented modification
```

```kotlin
// List access using index & extension functions(first, last)
val readOnlyShapes = listOf("triangle", "square", "circle")
println("The first item in the list is: ${readOnlyShapes[0]}")
println("The first item in the list is: ${readOnlyShapes.first()}")
println("The last item in the list is: ${readOnlyShapes.last()}")
```

```kotlin
// Number of items in list
val readOnlyShapes = listOf("triangle", "square", "circle")
println("This list has ${readOnlyShapes.count()} items")
```

```kotlin
// Check item is in list
val readOnlyShapes = listOf("triangle", "square", "circle")
println("circle" in readOnlyShapes)
```

```kotlin
// Adding or removing items from mutable list
val shapes: MutableList<String> = mutableListOf("triangle", "square", "circle")
shapes.add("hexagon")
println(shapes)

shapes.remove("hexagon")
println(shapes)
```

```kotlin
// Check item is in set
val readOnlyFruit = setOf("apple", "banana", "cherry", "cherry")
println("cherry" in readOnlyFruit)
```

```kotlin
// Adding or removing items from mutable set
val fruit: MutableSet<String> = mutableSetOf("apple", "banana", "cherry", "cherry")
fruit.add("jackfruit")
println(fruit)

fruit.remove("jackfruit")
println(fruit)
```

## Map
Mainly used when items are in key-value pairs (JSON)

```kotlin
// Read-only & mutable map(with type declaration)
val readOnlyJuiceMenu = mapOf("apple" to 100, "kiwi" to 190, "orange" to 67)
println(readOnlyJuiceMenu)

val juiceMenu: MutableMap<String, Int> = mutableMapOf("apple" to 100, "kiwi" to 190, "orange" to 67)
println(juiceMenu)
```

```kotlin
// Casting(prevent unwanted modifications to map)
val juiceMenu: MutableMap<String, Int> = mutableMapOf("apple" to 100, "kiwi" to 190, "orange" to 67)
val juiceMenuLocked: Map<String, Int> = juiceMenu

juiceMenu.put("tomato", 88) // Original still modifiable
juiceMenuLocked.put("carrot", 69) // Prevented modification
```

```kotlin
// Map access using index
val readOnlyJuiceMenu = mapOf("apple" to 100, "kiwi" to 190, "orange" to 67)
println("The value of apple juice is: ${readOnlyJuiceMenu["orange"]}")
println("The value of apple juice is: ${readOnlyJuiceMenu["banana"]}") // null when not in map
```

```kotlin
// Adding or removing items from mutable map
val juiceMenu: MutableMap<String, Int> = mutableMapOf("apple" to 100, "kiwi" to 190, "orange" to 67)
juiceMenu["coconut"] = 169 // Using index
juiceMenu.put("tomato", 88) // Using put
println(juiceMenu)

juiceMenu.remove("orange")
println(juiceMenu)
```

```kotlin
// Number of items in map
val readOnlyJuiceMenu = mapOf("apple" to 100, "kiwi" to 190, "orange" to 67)
println("This map has ${readOnlyJuiceMenu.count()} key to value pairs")
```

```kotlin
// Get collection of keys or value from map
val readOnlyJuiceMenu = mapOf("apple" to 100, "kiwi" to 190, "orange" to 67)
println(readOnlyJuiceMenu.keys)
println(readOnlyJuiceMenu.values)
```

```kotlin
// Check key or value is in map
val readOnlyJuiceMenu = mapOf("apple" to 100, "kiwi" to 190, "orange" to 67)
println(readOnlyJuiceMenu.containsKey("kiwi"))
println(readOnlyJuiceMenu.contains("kiwi"))
println("kiwi" in readOnlyJuiceMenu.keys)
println("kiwi" in readOnlyJuiceMenu)

println(readOnlyJuiceMenu.containsValue(190))
println(190 in readOnlyJuiceMenu.values)
```

## Conditions

```kotlin
// if condition
var d: Int
val check = true

if (check){
    d = 1
} else {
    d = 2
}

println(d)
```

```kotlin
// if as an expression
val a = 1
val b = 2

println(if (a > b) a else b)
```

```kotlin
// when condition as a statement (no return value)
val obj = "Hello"

when (obj) {
    "1" -> println("One")
    "Hello" -> println("Greeting")
    else -> println("Unknown")
}
```

```kotlin
// when condition as an expression (return value)
val obj = "Hello"

val result = when(obj) {
    "1" -> "One"
    "Hello" -> "Greeting"
    else -> "Unknown"
}

println(result)
```

```kotlin
// When condition without subject (no obj in parameter, not reccomended)
val trafficLightState = "Red"

val trafficAction = when {
    trafficLightState == "Green" -> "Go"
    trafficLightState == "Yellow" -> "Slow down"
    trafficLightState == "Red" -> "Stop"
    else -> "Malfunction"
}

println(trafficAction)
```

```kotlin
// Named arguments, enables swapping positions in function parameter signature and more readable
fun printMessageWithPrefix(message: String, prefix: String){
    println("[$prefix] $message")
}

printMessageWithPrefix(prefix = "Log", message = "Hello")

// Not allowed, once parameter is named, subsequent must be named
// printMessageWithPrefix(prefix = "Log", "Hello")
```

```kotlin
// Early return, guard clause?
val registeredUsernames = mutableListOf("john_doe", "jane_smith")
val registeredEmails = mutableListOf("john@example.com", "jane@example.com")

fun registerUser(username: String, email: String): String{
    if (username in registeredUsernames)
        return "Username already taken. Please choose a different username."

    if(email in registeredEmails)
        return "Email already registered. Please use a different email."

    registeredUsernames.add(username)
    registeredEmails.add(email)

    return "User registered successfully: $username"
}

println(registerUser("john_doe", "newjohn@example.com"))
println(registerUser("jason_tan", "john@example.com"))
println(registerUser("new_user", "newuser@example.com"))
```

## Lambda Expressions

```kotlin
// From function to lambda expression
fun baseUppercaseString(text: String): String {
    return text.uppercase()
}

val upperCaseString = { text: String -> text.uppercase() }

println(baseUppercaseString("hello"))
println(upperCaseString("hello"))
```

```kotlin
// Pass to another function
val numbers = listOf(1, -2, 3, -4, 5, -6)

// Direct pass
val positives = numbers.filter({x -> x > 0})

// Pass as parameter
val isNegative = { x: Int -> x < 0 }
val negatives = numbers.filter(isNegative)

// Trailing lambda in parameters can ignore parenthesis, can use "it"
val positivesTrailingLambda = numbers.filter {it > 0}

println(positives)
println(negatives)
println(positivesTrailingLambda)
```

```kotlin
// Using map
val numbers = listOf(1, -2, 3, -4, 5, -6)
val doubled = numbers.map {it * 2}

val triple = { x:Int -> x * 3}
val tripled = numbers.map(triple)

println(doubled)
println(tripled)
```

```kotlin
// Function types, normally Kotlin infers "(String) -> String" from right hand side
val upperCaseString: (String) -> String = { text -> text.uppercase() }

println(upperCaseString("hello"))
```

```kotlin
// Returning lambda expression from functions
fun toSeconds(time: String): (Int) -> Int = when (time){
    "hour" -> { value -> value * 60 * 60 }
    "minute" -> { value -> value * 60}
    "second" -> { value -> value }
    else -> { value -> value }
}

val timesInMinutes = listOf(2, 10, 15, 1)
val min2sec = toSeconds("minute")
val totalMinTimeInSeconds = timesInMinutes.map(min2sec).sum()
println("Total time is $totalMinTimeInSeconds secs")

val timesInHours = listOf(2, 8, 6, 1)
val hour2sec = toSeconds("hour")
val totalHourTimeInSeconds = timesInHours.map(hour2sec).sum()
println("Total time is $totalHourTimeInSeconds secs")
```

```kotlin
// Invoking lambda immediately (IIFE?)
println({ text: String -> text.uppercase() }("hello"))
```

```kotlin
// Trailing lambdas with additional parameters
val numList = listOf(1, 2, 3)

println(numList.fold(0, { x, item -> x + item }))
println(numList.fold(0) { x, item -> x + item })
```
